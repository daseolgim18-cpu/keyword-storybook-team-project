```python
import pandas as pd

# CSV 파일 읽기
df = pd.read_csv("/content/학생의_전자책_독서율_변화_추이_20241121154651.csv" )

df
```

```python
# 첫 번째 행 삭제 (index 0)
df = df.drop(index=0)

# 결과 확인
df.head()
```

```python
# 열 이름 바꾸기
df.columns = df.columns.str.replace("131018523991 통계분류(1)", "통계분류(1)")
df.columns = df.columns.str.replace("131018523991 통계분류(2)", "통계분류(2)")
df.columns = df.columns.str.replace("131018523991 통계분류(3)", "통계분류(3)")
df.columns = df.columns.str.replace("2Y2010 2010", "2010년")
df.columns = df.columns.str.replace("2Y2011 2011", "2011년")

# 결과 확인
df.head()
```

```python
# "131028523991.003 "와 같은 부분을 제거하고 "학교급별+성별"만 남기기
df['통계분류(1)'] = df['통계분류(1)'].str.replace(r'^\d+\.\d+ ', '', regex=True)

# 결과 확인
df[['통계분류(1)']].head()
```

```python
# "131028523991.003 "와 같은 부분을 제거하고 "초등학생, 중학생, 고등학생"만 남기기
df['통계분류(2)'] = df['통계분류(2)'].str.replace(r'^\d+\.\d+ ', '', regex=True)

# 결과 확인
df[['통계분류(2)']].head()
```

```python
# "131028523991.003 "와 같은 부분을 제거하고 "학교급별+성별"만 남기기
df['통계분류(3)'] = df['통계분류(3)'].str.replace(r'^\d+\.\d+ ', '', regex=True)

# 결과 확인
df[['통계분류(3)']].head()
```

```python
import matplotlib.pyplot as plt  # 그래프 그리기
import koreanize_matplotlib

# # 한글 깨짐 현상
# plt.rc('font', family='NanumBarunGothic')

# '2010년'과 '2011년' 데이터, '통계분류(2)' 열에서 인덱스 0~2의 데이터 추출
x_labels = df['통계분류(2)'].iloc[0:3]  # x축 값 (0~2 인덱스)
y_2010 = df['2010년'].iloc[0:3]  # '2010년' 데이터
y_2011 = df['2011년'].iloc[0:3]  # '2011년' 데이터

# 막대 그래프 그리기
fig, ax = plt.subplots(figsize=(8, 6))  # 그래프 크기 조정 (8x6)

# 막대 너비 설정
bar_width = 0.35

# x축 인덱스 생성 (0, 1, 2)
index = range(len(x_labels))

# 2010년과 2011년 데이터를 각각 다른 색으로 그리기
bars_2010 = ax.bar(index, y_2010, bar_width, label='2010년', color='blue')
bars_2011 = ax.bar([i + bar_width for i in index], y_2011, bar_width, label='2011년', color='orange')

# x축 레이블 설정
ax.set_xticks([i + bar_width / 2 for i in index])  # 막대 그래프 중간에 레이블 위치
ax.set_xticklabels(x_labels)  # '통계분류(2)' 데이터로 x축 레이블 설정

# 제목과 레이블 설정
ax.set_title('학생의_전자책_독서율_변화_추이', fontsize=14)  # 제목 설정
ax.set_xlabel('통계분류(2)', fontsize=12)  # x축 라벨 크기 설정
ax.set_ylabel('독서율', fontsize=12)  # y축 라벨 크기 설정

# 범례 추가
ax.legend()

# 그래프 표시
plt.tight_layout()  # 레이아웃을 자동으로 조정하여 겹침 방지
plt.show()
```
