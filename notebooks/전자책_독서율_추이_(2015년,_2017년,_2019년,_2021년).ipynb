```python
import pandas as pd

# 데이터 읽기
df2 = pd.read_excel( '/content/전자책 독서율 추이 (2015년, 2017년, 2019년, 2021년, 2023년).xlsx')
df2
```

```python
import matplotlib.pyplot as plt
import koreanize_matplotlib

# Extract data for visualization from the second DataFrame (e-book reading trend)
years = ['2015년', '2017년', '2019년', '2021년', '2023년']
values = df2.iloc[0, 1:].values.astype(float)

# 증감률 계산 (각 연도 간 증감률)
growth_rate = [(values[i] - values[i-1]) / values[i-1] * 100 if i > 0 else 0 for i in range(len(values))]

# 그래프 생성
fig, ax1 = plt.subplots(figsize=(10, 6))

# 막대 그래프 그리기
ax1.bar(years, values, alpha=0.7, label='독서율', color='skyblue')

# 선 그래프 그리기 (증감률을 나타내는 선 그래프)
ax1.plot(years, values, marker='o', color='b', linestyle='-', linewidth=2, markersize=6, label='독서율 추이')

# 그래프 설정
ax1.set_xlabel('연도')
ax1.set_ylabel('독서율 (%)')
ax1.set_title('전자책 독서율 추이')
ax1.set_xticks(years)
ax1.grid(True)

# 레전드 추가
ax1.legend()

# 그래프 출력
plt.tight_layout()
plt.show()
```
