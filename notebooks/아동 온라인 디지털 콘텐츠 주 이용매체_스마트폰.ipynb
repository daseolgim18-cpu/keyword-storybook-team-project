```python
import pandas as pd

# 데이터 읽기
df = pd.read_excel('/content/아동 온라인 디지털 콘텐츠 주 이용매체_스마트폰.xlsx')
df
```

```python
import matplotlib.pyplot as plt
import koreanize_matplotlib

# 막대 그래프 그리기
df.set_index('나이').plot(kind='bar', figsize=(10, 6))

# 그래프 제목과 레이블 설정
plt.title('0세~19세 스마트폰 온라인 디지털 콘텐츠 주 이용 통계', fontsize=14)
plt.xlabel('나이 구간', fontsize=12)
plt.ylabel('값', fontsize=12)
plt.xticks(rotation=0)
plt.legend(title='항목', fontsize=10)

# 그래프 표시
plt.tight_layout()
plt.show()
```
